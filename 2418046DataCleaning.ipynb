CELL 1 — Upload File Excel

```python
from google.colab import files

uploaded = files.upload()
```

CELL 2 — Import Library

```python
import pandas as pd
import re
```

CELL 3 — Membaca Excel

```python
# Mengambil nama file Excel yang di-upload
nama_file = list(uploaded.keys())[0]

# Membaca Excel tanpa langsung menentukan nama kolom
data_awal = pd.read_excel(nama_file, header=None)

# Menampilkan 10 baris pertama
data_awal.head(10)
```

CELL 4 — Mencari Header Secara Otomatis

```python
# Mencari baris yang memiliki tulisan "NIM"
baris_header = None

for i in range(len(data_awal)):
    isi_baris = data_awal.iloc[i].astype(str).str.strip().tolist()

    if "NIM" in isi_baris:
        baris_header = i
        break

print("Baris header ditemukan pada baris:", baris_header)
```

CELL 5 — Membuat Dataset

```python
# Membaca kembali Excel dengan header yang sudah ditemukan
df = pd.read_excel(nama_file, header=baris_header)

# Membersihkan nama kolom
df.columns = (
    df.columns
    .astype(str)
    .str.replace("\n", " ", regex=False)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

# Menampilkan nama kolom
print("Nama kolom:")
print(df.columns.tolist())
```

CELL 6 — Melihat Dataset Awal

```python
print("Jumlah data:", len(df))
print("Jumlah kolom:", len(df.columns))

df.head(10)
```

CELL 7 — Mengecek Data Kosong

```python
print("Jumlah data kosong:")
print(df.isnull().sum())
```

CELL 8 — Mengecek Data Duplikat

```python
print("Jumlah data duplikat:", df.duplicated().sum())
```

CELL 9 — Cleaning Nama Mahasiswa

```python
df["Nama Lengkap Mahasiswa"] = (
    df["Nama Lengkap Mahasiswa"]
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Nama mahasiswa setelah dibersihkan:")
print(df[["NIM", "Nama Lengkap Mahasiswa"]].head(10))
```

CELL 10 — Cleaning Program Studi

```python
df["Program Studi & Fakultas"] = (
    df["Program Studi & Fakultas"]
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Program studi setelah dibersihkan:")
print(df[["NIM", "Program Studi & Fakultas"]].head(10))
```

CELL 11 — Cleaning IPK

```python
df["IPK Semester 1"] = (
    df["IPK Semester 1"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

df["IPK Semester 1"] = pd.to_numeric(
    df["IPK Semester 1"],
    errors="coerce"
)

print("IPK setelah dibersihkan:")
print(df[["NIM", "IPK Semester 1"]].head(10))
```

CELL 12 — Cleaning Status

```python
def bersihkan_status(status):
    status = str(status).strip().lower()

    if status in ["1", "true", "ya"]:
        return "Ya"
    elif status in ["0", "false", "tidak"]:
        return "Tidak"
    else:
        return "Tidak diketahui"


df["Status Beasiswa / Aktif"] = (
    df["Status Beasiswa / Aktif"]
    .apply(bersihkan_status)
)

print("Status setelah dibersihkan:")
print(df[["NIM", "Status Beasiswa / Aktif"]].head(10))
```

CELL 13 — Cleaning Nomor Telepon

```python
def bersihkan_telepon(no):
    no = str(no)

    # Menghapus spasi, tanda +, tanda -, dan karakter lain
    no = re.sub(r"\D", "", no)

    # Jika menggunakan kode negara 62, ubah menjadi 0
    if no.startswith("62"):
        no = "0" + no[2:]

    return no


df["No. Telepon / WA"] = (
    df["No. Telepon / WA"]
    .apply(bersihkan_telepon)
)

print("Nomor telepon setelah dibersihkan:")
print(df[["NIM", "No. Telepon / WA"]].head(10))
```

CELL 14 — Cleaning Tanggal Lahir

```python
def bersihkan_tanggal(tanggal):
    tanggal = str(tanggal).strip()

    # Format YYYY/MM/DD atau YYYY-MM-DD
    if re.match(r"^\d{4}[/\-]\d{1,2}[/\-]\d{1,2}$", tanggal):
        return pd.to_datetime(tanggal, errors="coerce")

    # Format seperti 03/15/2006 atau 08/30/06
    if "/" in tanggal:
        return pd.to_datetime(tanggal, errors="coerce", format="mixed")

    # Format seperti 12-05-2005 atau 28-02-06
    if "-" in tanggal:
        return pd.to_datetime(tanggal, errors="coerce", dayfirst=True)

    # Format tulisan seperti 14 November 2005
    return pd.to_datetime(tanggal, errors="coerce", dayfirst=True)


df["Tanggal Lahir"] = df["Tanggal Lahir"].apply(bersihkan_tanggal)

# Mengubah menjadi format YYYY-MM-DD
df["Tanggal Lahir"] = df["Tanggal Lahir"].dt.strftime("%Y-%m-%d")

print("Tanggal lahir setelah dibersihkan:")
print(df[["NIM", "Tanggal Lahir"]].head(10))
```
